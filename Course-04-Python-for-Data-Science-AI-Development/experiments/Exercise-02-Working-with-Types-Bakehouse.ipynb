## Immanuel's Bakehouse Application

This experiment applies Python data types to bakery
costing, production and sales scenarios.

```python
# Bakehouse - Python Data Types

flour_cost = 60.50
chocolate_cost = 250.00
batch_quantity = 6
selling_price = 600.00
product_name = "Chocolate Brownie"
product_available = True

print(flour_cost)
print(chocolate_cost)
print(batch_quantity)
print(selling_price)
print(product_name)
print(product_available)
```

```python
print(type(flour_cost))
print(type(chocolate_cost))
print(type(batch_quantity))
print(type(selling_price))
print(type(product_name))
print(type(product_available))
```

```python
selling_price_text =str( selling_price)
print(selling_price_text)
print(type(selling_price_text))
```

```python
batch_quantity_decimal = float(batch_quantity)
print(batch_quantity_decimal)
print(type(batch_quantity_decimal))
```

## 1. Batch Cost

```python
flour_cost = 450
sugar_cost = 180
butter_cost = 320
labor_cost = 250
packaging_cost = 100
#Calculate the total batch cost: 
batch_cost = (flour_cost + sugar_cost+ butter_cost+labor_cost+packaging_cost)
print(f"Batch_Cost: ₹{batch_cost: .2f}")
```

## 2. Unt Cost

```python
units_produced = 50
unit_cost = batch_cost/units_produced
print(f"unit_Cost: ₹{unit_cost: .2f}")
```

## Selling Price and Margin

```python
selling_price = 40
margin = selling_price - unit_cost
print(f"Margin: ₹{margin:.2f}")
```

```python
#Calculate the margin percentage:
margin_percentage = (margin / selling_price) * 100

print(f"Margin Percentage: {margin_percentage:.2f}%")
```

## Combined Bakehouse Calculation

```python
flour_cost = 450
sugar_cost = 180
butter_cost = 320
labor_cost = 250
packaging_cost = 100
units_produced = 50
selling_price = 40
batch_cost = (
    flour_cost
    + sugar_cost
    + butter_cost
    + labor_cost
    + packaging_cost
)
unit_cost = batch_cost / units_produced
margin = selling_price - unit_cost
margin_percentage = (margin / selling_price) * 100
print(f"Batch Cost: ₹{batch_cost:.2f}")
print(f"Unit Cost: ₹{unit_cost:.2f}")
print(f"Margin: ₹{margin:.2f}")
print(f"Margin Percentage: {margin_percentage:.2f}%")
```
